```python
from datetime import date, timedelta
import sqlite3
import pandas as pd
from pybaseball import cache, statcast
```

```python
cache.enable()

database_path = "statcast.sqlite"
start_date = date(2016, 1, 1)
end_date = date(2026, 9, 1)
first_chunk = True
```

```python
with sqlite3.connect(database_path) as connection:
    current_date = start_date

    while current_date <= end_date:
        chunk_end = min(current_date + timedelta(days=6), end_date)
        print(f"Downloading {current_date} through {chunk_end}")

        chunk = statcast(
            current_date.isoformat(),
            chunk_end.isoformat()
        )
        if chunk is None or chunk.empty or len(chunk.columns) == 0:
            print("No data found; skipping chunk.")
            current_date = chunk_end + timedelta(days=1)
            continue
        chunk.to_sql(
            "statcast",
            connection,
            if_exists="replace" if first_chunk else "append",
            index=False,
            chunksize=1000,
        )

        first_chunk = False
        current_date = chunk_end + timedelta(days=1)

    connection.execute(
        "CREATE INDEX IF NOT EXISTS idx_statcast_game_date "
        "ON statcast(game_date)"
    )
    connection.execute(
        "CREATE INDEX IF NOT EXISTS idx_statcast_batter "
        "ON statcast(batter)"
    )
```
